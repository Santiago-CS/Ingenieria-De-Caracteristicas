# Relaciones lineales

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
```

```python
# Ayudín
'''Argumentos de 'minimize'
fun: función objetivo a minimizar fun(x,*args) -> float
x0: parámetros a encontrar [a0,a1]
args: argumentos extras que necesita fun (x,y)
'''
# minimize?
```

Definimos el criterio de optimización:
$$
\frac{1}{N}\sum^N(a_0+a_1x-y)^2,
$$
donde $x$ y $y$ son los pares de observaciones, $N$ la cantidad de observaciones y los coeficientes $a_0$ y $a_1$, los parámetros a encontrar.

```python
# Encontramos la relación entre las variables y='ash' y x='alcalinity_of_ash'
def ecm(a,x,y):
    return np.mean((a[0]+a[1]*x-y)**2)
```

```python
# Para agregar una celda es con 'esc'+'b' o 'esc'+'a'
#
a_inicial=np.random.rand(2) # valores aleatorios de dimensión 2
solucion=minimize(ecm,a_inicial,args=(df_wine['alcalinity_of_ash'],df_wine['ash']))
solucion
```

## Usando SKLearn

```python
from sklearn.linear_model import LinearRegression
```

```python
df_wine['alcalinity_of_ash'].shape # Con un corchete
```

```python
df_wine[['alcalinity_of_ash']].shape # Con dos corchetes
```

```python
# Definimos un objeto que sea de tipo 'LinearRegression'
lin=LinearRegression()
lin.fit(df_wine[['alcalinity_of_ash']],df_wine['ash'])
```

```python
# Que tan bien se ajustó
lin.score(df_wine[['alcalinity_of_ash']],df_wine['ash']) # R^2
```

Ahora con dos variables independientes (`alcalinity_of_ash` y `alcohol`)

```python
lin.fit(df_wine[['alcalinity_of_ash','alcohol']],df_wine['ash'])
# Que tan bien se ajustó
lin.score(df_wine[['alcalinity_of_ash','alcohol']],df_wine['ash']) # R^2
```

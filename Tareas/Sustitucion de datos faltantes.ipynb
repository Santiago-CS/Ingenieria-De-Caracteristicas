# Análisis de Variables y Tratamiento de Valores Faltantes

Tras explorar el archivo `tratamiento_de_valores_faltantes-1.csv`, se identifican las siguientes características para cada variable:

* **x1 (Numérica Continua):** Presenta una distribución relativamente simétrica (sesgo de 0.25), pero **contiene 7 datos atípicos** significativos (valores por debajo de -1.31 y por encima de 24.88). Tiene 18 valores faltantes.
* **x2 (Categórica nominal):** Representa tipos de comida (Mexicano, Comida rápida, etc.). Su categoría más frecuente (moda) es "Mexicano". Tiene 20 valores faltantes.
* **x3 (Numérica Continua):** Presenta una distribución simétrica (sesgo de 0.10) y **no presenta datos atípicos**. Tiene 14 valores faltantes.
* **x4 (Numérica Continua):** Presenta una distribución simétrica (sesgo de 0.11) y **no presenta datos atípicos**. Tiene 17 valores faltantes.

**Estrategia de Imputación (Sustitución):**
1. **x1:** Al tener datos atípicos, la **mediana** es la mejor opción, ya que es una medida robusta que no se deja afectar por los valores extremos.
2. **x2:** Por ser categórica, utilizaremos la **moda** (el valor más frecuente).
3. **x3 y x4:** Al ser numéricas con distribución simétrica y sin atípicos, utilizaremos la **media** (promedio).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cargar el archivo
df = pd.read_csv('C:/Users/santi/OneDrive - ITESO/ITESO/Semestre 3/Ingenieria de Caracteristicas/Ingenieria-De-Caracteristicas/Datos/tratamiento_de_valores_faltantes-1.csv')

# 2. Visualizar la distribución y detectar datos atípicos (Outliers) en numéricas
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
sns.boxplot(y=df['x1'], ax=axes[0]).set_title('x1 (Atípicos presentes)')
sns.boxplot(y=df['x3'], ax=axes[1]).set_title('x3 (Sin atípicos)')
sns.boxplot(y=df['x4'], ax=axes[2]).set_title('x4 (Sin atípicos)')
plt.tight_layout()
plt.show()

# Resumen de valores faltantes iniciales
print("Valores faltantes ANTES de la imputación:\n", df.isnull().sum())
```

```python
# 1. Imputar x1 (Numérica con atípicos) -> Mediana
mediana_x1 = df['x1'].median()
df['x1'] = df['x1'].fillna(mediana_x1)

# 2. Imputar x2 (Categórica) -> Moda
moda_x2 = df['x2'].mode()[0]
df['x2'] = df['x2'].fillna(moda_x2)

# 3. Imputar x3 (Numérica sin atípicos) -> Media
media_x3 = df['x3'].mean()
df['x3'] = df['x3'].fillna(media_x3)

# 4. Imputar x4 (Numérica sin atípicos) -> Media
media_x4 = df['x4'].mean()
df['x4'] = df['x4'].fillna(media_x4)

# Comprobación de que no quedan valores nulos
print("\nValores faltantes DESPUÉS de la imputación:\n", df.isnull().sum())
```
